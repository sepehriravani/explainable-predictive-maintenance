## 02 — Data Preprocessing

In this notebook, I prepare the dataset for machine learning.

```python
!pip install -q ucimlrepo
```

```python
from ucimlrepo import fetch_ucirepo
```

```python
data = fetch_ucirepo(id=601)

X = data.data.features
y = data.data.targets
```

```python
target = y["Machine failure"]
```

```python
df = X.copy()
df["Machine failure"] = target
```

```python
df.head()
```

```python
features = df.drop("Machine failure", axis=1)
target = df["Machine failure"]
```

```python
features.head()
```

```python
import pandas as pd
```

```python
features_encoded = pd.get_dummies(
    features,
    columns=["Type"],
    dtype=int
)
```

```python
features_encoded.head()
```

```python
features_encoded.columns
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    features_encoded,
    target,
    test_size=0.2,
    random_state=42,
    stratify=target
)
```

```python
print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)
```

```python
print("Train failure rate:")
print(y_train.value_counts(normalize=True) * 100)

print("\nTest failure rate:")
print(y_test.value_counts(normalize=True) * 100)
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
numeric_columns = [
    "Air temperature",
    "Process temperature",
    "Rotational speed",
    "Torque",
    "Tool wear"
]
```

```python
X_train_scaled = X_train.copy()
X_test_scaled = X_test.copy()
```

```python
scaler = StandardScaler()

X_train_scaled[numeric_columns] = scaler.fit_transform(
    X_train[numeric_columns]
)
```

```python
X_test_scaled[numeric_columns] = scaler.transform(
    X_test[numeric_columns]
)
```
